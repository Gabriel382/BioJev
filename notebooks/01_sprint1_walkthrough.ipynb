```python
subprocess.run(
    [
        sys.executable,
        "scripts/evaluate.py",
        "--model-config", "configs/models/openjev.yaml",
        "--dataset", "bionli",
        "--split", "test",
        "--seed", "42",
        "--max-examples", "50",
    ],
    check=True,
)

```
